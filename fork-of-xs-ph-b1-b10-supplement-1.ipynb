```python
from pathlib import Path
import subprocess
import sys
import os
import json

# All supplement files are inside this Kaggle Dataset directory.
DATASET_DIR = Path(
    "/kaggle/input/datasets/tariqkanaan/xs-ph-b1-b10-supplement-package/XS_PH_B1_B10_SUPPLEMENT_PACKAGE"
)

PKG = DATASET_DIR

if not PKG.exists():
    raise FileNotFoundError(
        f"Supplement package directory was not found: {PKG}. "
        "Check that the Kaggle Dataset is attached and that the folder name is correct."
    )

required_files = [
    "run_supplement_b1_b10.py",
    "run_xs_study_v2.py",
    "analyze_supplement_b1_b10.py",
    "validate_supplement.py",
    "requirements.txt",
    "supplement_protocol.json",
]

missing = [
    name for name in required_files
    if not (PKG / name).is_file()
]

if missing:
    raise FileNotFoundError(
        f"Supplement package is incomplete at {PKG}; missing files: {missing}"
    )

OUT = Path("/kaggle/working/XS_PH_B1_B10_SUPPLEMENT")
OUT.mkdir(parents=True, exist_ok=True)

print("Package:", PKG)
print("Required files: OK")
print("Output:", OUT)
```

```python
subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "-q",
        "-r",
        str(PKG / "requirements.txt")
    ],
    check=True
)

print("Required packages installed successfully.")
```

```python
subprocess.run(
    [
        sys.executable,
        str(PKG / "run_supplement_b1_b10.py"),
        "--outroot",
        str(OUT),
        "--run",
        "B2_rho_cap"
    ],
    check=True
)
```

```python
# Analyze all completed supplementary experiments.

subprocess.run(
    [
        sys.executable,
        str(PKG / "analyze_supplement_b1_b10.py"),
        "--root",
        str(OUT)
    ],
    check=True
)

# Validate the supplementary results.

subprocess.run(
    [
        sys.executable,
        str(PKG / "validate_supplement.py"),
        "--root",
        str(OUT)
    ],
    check=True
)

print("Supplementary analysis and validation completed.")
```
